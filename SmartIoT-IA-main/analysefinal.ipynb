```python
# ============================================
# VERIFICATION DES FICHIERS EXISTANTS
# ============================================

import os
import json

print(" VERIFICATION DES FICHIERS DE RESULTATS IoT")
print("=" * 70)

# Liste de tous les fichiers dans le dossier
fichiers_existants = sorted(os.listdir('.'))
fichiers_json = sorted([f for f in fichiers_existants if f.endswith('.json')])

print(" FICHIERS JSON TROUVÉS DANS LE DOSSIER :")
print("-" * 70)

for i, f in enumerate(fichiers_json, 1):
    if os.path.exists(f):
        size = os.path.getsize(f) / 1024  # Taille en KB
        print(f"  {i:2}.  {f:50} ({size:.1f} KB)")
    else:
        print(f"  {i:2}.  {f:50} (ABSENT)")

# -----------------------------------------------------------------
print("\n VERIFICATION DES 7 FICHIERS REQUIS POUR L'ANALYSE COMPARATIVE")
print("=" * 70)

# Les 7 fichiers requis selon ton image
fichiers_requis = [
    "resultats_garage_door.json",
    "resultats_Train_Test_IoT_Fridge.json",
    "resultats_Train_Test_IoT_GPS_Tracker.json",
    "resultats_Train_Test_IoT_Modbus.json",
    "resultats_Train_Test_IoT_Motion_Light.json",
    "resultats_Train_Test_IoT_Thermostat.json",
    "resultats_Train_Test_IoT_Weather.json"
]

print("\nSTATUT DES 7 FICHIERS IoT :")
print("-" * 70)

status = {}
for fichier in fichiers_requis:
    if os.path.exists(fichier):
        size = os.path.getsize(fichier) / 1024
        status[fichier] = {'present': True, 'size_kb': size}
        print(f" {fichier:60} ✓ ({size:.1f} KB)")
    else:
        status[fichier] = {'present': False}
        print(f" {fichier:60} ✗ MANQUANT")

# -----------------------------------------------------------------
print("\n CONTENU ET VALIDATION DES FICHIERS JSON")
print("=" * 70)

for fichier in fichiers_requis:
    if fichier in status and status[fichier]['present']:
        print(f"\n ANALYSE DE '{fichier}' :")
        print("-" * 50)
        
        try:
            with open(fichier, 'r', encoding='utf-8') as f:
                data = json.load(f)
            
            # Vérification basique
            print(f"  • Format JSON : ✓ VALIDE")
            
            # Informations clés
            if 'dataset' in data:
                print(f"  • Dataset : {data['dataset']}")
            
            if 'dimensions' in data:
                print(f"  • Dimensions : {data['dimensions'][0]} lignes × {data['dimensions'][1]} colonnes")
            
            if 'distribution_classes' in data:
                print(f"  • Distribution : {data['distribution_classes']}")
            
            if 'resultats_comparatifs' in data:
                print(f"  • Modèles évalués : {len(data['resultats_comparatifs'])}")
                for modele in data['resultats_comparatifs']:
                    print(f"    - {modele['Modele']}: F1={modele.get('F1-Score', 'N/A'):.3f}")
        
        except json.JSONDecodeError as e:
            print(f"  • Format JSON :  ERREUR - {e}")
        except Exception as e:
            print(f"  • Erreur lecture : {str(e)[:100]}...")

# -----------------------------------------------------------------
print("\n STATISTIQUES GLOBALES")
print("=" * 70)

fichiers_presents = sum(1 for f in fichiers_requis if f in status and status[f]['present'])
print(f"Fichiers présents : {fichiers_presents} sur {len(fichiers_requis)}")

if fichiers_presents == len(fichiers_requis):
    print(" TOUS LES FICHIERS SONT PRÉSENTS ! Prêt pour l'analyse comparative.")
    
    # Calcul des tailles
    taille_totale = sum(status[f]['size_kb'] for f in fichiers_requis if f in status and status[f]['present'])
    print(f" Taille totale des données : {taille_totale:.1f} KB")
    
    # Charger tous les datasets pour vérifier la cohérence
    print("\n COMPATIBILITÉ DES DATASETS :")
    print("-" * 40)
    
    datasets_info = {}
    for fichier in fichiers_requis:
        try:
            with open(fichier, 'r') as f:
                data = json.load(f)
                
            dataset_name = data.get('dataset', 'Inconnu')
            dimensions = data.get('dimensions', [0, 0])
            n_lignes = dimensions[0] if isinstance(dimensions, list) and len(dimensions) > 0 else 0
            
            datasets_info[fichier] = {
                'name': dataset_name,
                'lignes': n_lignes,
                'modeles': len(data.get('resultats_comparatifs', []))
            }
            
            print(f"  {os.path.basename(dataset_name):25} → {n_lignes:>8,} lignes | {len(data.get('resultats_comparatifs', []))} modèles")
            
        except:
            datasets_info[fichier] = {'name': 'Erreur', 'lignes': 0, 'modeles': 0}
    
else:
    print(f"  Attention : {len(fichiers_requis) - fichiers_presents} fichiers manquants")
    print("Fichiers manquants :")
    for fichier in fichiers_requis:
        if fichier not in status or not status[fichier]['present']:
            print(f"  • {fichier}")

# -----------------------------------------------------------------
print("\n PRÉPARATION POUR L'ANALYSE COMPARATIVE")
print("=" * 70)

if fichiers_presents == len(fichiers_requis):
    print("""
ÉTAPES SUIVANTES :
1. Créer un notebook d'analyse comparative des 7 appareils
2. Comparer les performances des algorithmes sur chaque dataset
3. Identifier le meilleur algorithme par type d'appareil IoT
4. Générer des visualisations comparatives
5. Produire un rapport final avec recommandations

STRUCTURE DES DONNÉES DISPONIBLES :
• 7 datasets IoT différents
• 3 algorithmes par dataset (KNN, Arbre, SVM)
• Métriques complètes pour chaque combinaison
• Tous les fichiers au même format JSON
""")
else:
    print("""
ACTION REQUISE :
1. Exécuter les codes Python manquants pour générer les fichiers JSON
2. Vérifier les noms des fichiers générés
3. Relancer cette vérification

LES CODES À EXÉCUTER (si manquants) :
• projet_ddos.py (Garage Door) -  EXISTE
• projet_fridge_corrige.py (Fridge)
• projet_motion_light_corrige.py (Motion Light)
• projet_thermostat_corrige.py (Thermostat)
• gps_trackers.ipynb (GPS Tracker)
• modbus.ipynb (Modbus)
• Train_Test_IoT_Weather.ipynb (Weather)
""")

print("\n" + "="*70)
print("VÉRIFICATION TERMINÉE")
print("="*70)
```
